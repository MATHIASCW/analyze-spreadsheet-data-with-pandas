Data cleaning


```python
df.drop_duplicates(inplace=True)
```

Rename Columns 

```python
df.rename(columns={"Name": "Title", "Year": "Publication Year", "User Rating": "Rating"}, inplace=True)

print(df.columns)
```

Edit data Price with float

```python
print(df.dtypes)
print(df.head())
```

```python
df["Price"] =  df["Price"].astype(float)
print(df.dtypes)
print(df.head()) 
```

Analyzing Author Popularity

```python
Author_Counts = df["Author"].value_counts()
print(Author_Counts)
```

Avg rating by genre

```python
avg_rg_genre = df.groupby("Genre")["Rating"].mean()
print(avg_rg_genre)
```

Results

```python
Author_Counts.head(10).to_csv("top_authors.csv")

avg_rg_genre.to_csv("avg_rating_by_genre.csv")
```
